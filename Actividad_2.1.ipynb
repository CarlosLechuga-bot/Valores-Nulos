```python
import pandas as pd
df =pd.read_csv("Citas_Digital_Filtrado.csv")
df.columns = df.columns.str.strip()

print("Valores nulos por columna:")
print(df.isnull().sum())

print("\nTAMAÑO DEL DATAFRAME:")
print(df.shape)
```

```python
#Tratatamiento de variables categóricas usando la moda
columnas_moda = ["BDC",
    "Estatus de Lead",
    "Asesor Asignado"]
for columna in columnas_moda: df[columna] = df[columna].fillna(df[columna].mode()[0])
#Tratamiento del nombre del cliente
df["Nombre Cliente"] = df["Nombre Cliente"].fillna("No registrado")
#Tratamiento de la fecha
df["Fecha"] = df["Fecha"].fillna("No registrado")
#Tratamiento de potencial de compra usando la mediana
df["Potencial de compra"] = pd.to_numeric(
    df["Potencial de compra"], 
    errors="coerce"
) 
df["Potencial de compra"] = df["Potencial de compra"].fillna(
    df["Potencial de compra"].median()
)

# Tratamiento de variables binarias usando la moda
columnas_binarias = [
    "PDM",
    "SDC",
    "Venta"
]

for columna in columnas_binarias:
    df[columna] = df[columna].fillna(df[columna].mode()[0])

# Tratamiento de preguntas abiertas
columnas_preguntas = [
    "¿Por que no ha visitado la agencia?",
    "¿Por que no hay solicitud de credito?",
    "¿Por que no hubo prueba de manejo ?",
    "¿Por que no hubo proceso wow?",
    "¿Por que se asigno antes de vistar piso?"
]
for columna in columnas_preguntas:
    df[columna] = df[columna].fillna("No especificado")

# Mostrar valores nulos después del tratamiento
print("\nVALORES NULOS DESPUÉS DEL TRATAMIENTO")
print(df.isnull().sum())
print("\nTIPO DE DATOS:")
print(df.dtypes)

print("\nEJEMPLOS DE FECHA:")
print(df["Fecha"].head(10))


df.to_csv("Citas_Digital_Tratado.csv", index=False)

print("\nArchivo final guardado como: Citas_Digital_Tratado.csv")
```
